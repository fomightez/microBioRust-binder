```python
import microbiorust as mb
collection = mb.gbk_to_faa("GCF_000210475.1_ASM21047v1_genomic.gbff")
print(collection)

#FaaCollection is a Rust-based structure we can convert into other python
#formats for ease of use
```

```python
#first 200 characters of the json string
print(collection.to_json()[:200])
```

```python
#access the json as a dictionary
import json
data_list = json.loads(collection.to_json())
first_key = list(data_list.keys())[0]
print(json.dumps(data_list[first_key], indent=4))
```

```python
#or grab all the records with parse
record_collection = mb.parse_gbk("GCF_000210475.1_ASM21047v1_genomic.gbff")

print(record_collection)
#the RecordCollection is a per-contig structure indexed by source key
#this genome file has one chromosome and 4 plasmids with separate source keys
```

```python
#access the features for parsing
for key in record_collection:
  print(key)

#the last key is a plasmid
#access the annotation of that plasmid
feature_collection = record_collection["NC_017633"].features()
#to access the sequences you can use .sequences() in the same way

#first key of feature_collection is the first coding sequence (CDS) on plasmid
print(feature_collection.keys()[0]) #the locus_tag of that CDS

#first key items
print(feature_collection.items()[0]) #returns key (locus_tag) and values
#first key values
print(feature_collection.values()[0]) #returns just values as a FeatureInfo

#note: Some() is a Rust Option string value
#extras could be used for parsing unsual labels, <i>e.g.</i> old_locus_tag
```
